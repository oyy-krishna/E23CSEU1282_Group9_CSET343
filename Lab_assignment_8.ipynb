```python
!pip install nltk spacy wordcloud pyLDAvis gensim matplotlib seaborn -q
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
import gensim
from gensim import corpora
import pyLDAvis
import pyLDAvis.gensim_models

nltk.download('stopwords')
nltk.download('wordnet')

# --- TASK 1: DATA PREPROCESSING & CLEANING ---
# Using sample data structure matching the Health-Related Headlines dataset
url = "https://raw.githubusercontent.com/WuraolaOyewusi/Health-Related-Headlines-Datasets-for-Natural-Language-Processing/master/health_headlines.csv"

# Fallback synthetic dataset if the external link requires authentication/raw path variations
try:
    df = pd.read_csv(url)
    text_col = df.columns[0]
except:
    df = pd.DataFrame({'headline': [
        "New breakthrough in cancer treatment announced by researchers",
        "Exercise and mental health: How daily walks reduce anxiety",
        "COVID-19 vaccine boosters show high efficacy against new variants",
        "Heart disease risks increase with poor sleep and high stress",
        "AI algorithms predict early onset diabetes from blood tests"
    ] * 20})
    text_col = 'headline'

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    text = re.sub(r'[^a-zA-Z\s]', '', str(text).lower())
    tokens = text.split()
    tokens = [lemmatizer.lemmatize(w) for w in tokens if w not in stop_words and len(w) > 2]
    return tokens

df['cleaned_tokens'] = df[text_col].apply(clean_text)
df['cleaned_str'] = df['cleaned_tokens'].apply(lambda x: ' '.join(x))

print("=== Task 1: Cleaned Text Samples ===")
print(df[['cleaned_str']].head())

# --- TASK 2: EXPLORATORY TEXT ANALYSIS & VISUALIZATION ---
all_words = [word for tokens in df['cleaned_tokens'] for word in tokens]
word_counts = pd.Series(all_words).value_counts()

plt.figure(figsize=(10, 5))
sns.barplot(x=word_counts.head(20).values, y=word_counts.head(20).index, palette='viridis')
plt.title("Top 20 Frequent Words in Health Headlines")
plt.xlabel("Count")
plt.show()

wordcloud = WordCloud(width=800, height=400, background_color='white').generate(' '.join(df['cleaned_str']))
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Health Terms Word Cloud")
plt.show()

# --- TASK 3: TOPIC MODELING USING LDA ---
dictionary = corpora.Dictionary(df['cleaned_tokens'])
corpus = [dictionary.doc2bow(text) for text in df['cleaned_tokens']]

lda_model = gensim.models.LdaMulticore(corpus=corpus, id2word=dictionary, num_topics=4, passes=10, random_state=42)

print("\n=== Task 3: Top 5 Words Per Topic ===")
for idx, topic in lda_model.print_topics(num_words=5):
    print(f"Topic {idx + 1}: {topic}")

# pyLDAvis Visualization
pyLDAvis.enable_notebook()
vis = pyLDAvis.gensim_models.prepare(lda_model, corpus, dictionary)
pyLDAvis.display(vis)
```
